```python
from optimal_clusters import evaluate_clustering_with_embeddings, generate_embedding_combinations
import pandas as pd
from sklearn.preprocessing import MinMaxScaler
import pickle, json
import numpy as np
from scipy.sparse import csr_matrix
from structures.deck import Deck
from structures.similarity import Similarity
```

```python
## Carica matrice dati e mazzi
def csr_load(path: str="cache/decks_csr.npz"):
    z = np.load(path, allow_pickle=True)
    X = csr_matrix(
        (z["data"], z["indices"], z["indptr"]),
        shape=tuple(z["shape"])
    )
    return X, z["deck_ids"], z["card_ids"]

# Carica i dati
X, deck_ids, card_ids = csr_load()

with open("cache/cards.json", encoding="utf-8") as f:
    encyclopedia = json.load(f)

with open("cache/decks.pkl", "rb") as f:
    decks = pickle.load(f)
# Crea i mazzi
# decks = [
#     Deck.from_sparse_row(
#         X.getrow(i), 
#         card_ids=card_ids, 
#         cards_encyclopedia=encyclopedia,
#         name=str(deck_ids[i])
#     )
#     for i in range(X.shape[0])
# ]

print(f"Caricati {len(decks)} mazzi")
```

```python
### Esecuzione della Valutazione

# EMBEDDINGS
embeddings_to_evaluate = generate_embedding_combinations([
    Deck.Embedding.ATTACKS_DMG,
    Deck.Embedding.ATTACKS_COSTS,
    Deck.Embedding.RESISTANCE,
    # Deck.Embedding.PKMN_TYPES,
    Deck.Embedding.SUPERTYPES,
    Deck.Embedding.KEYWORDS,
    # Deck.Embedding.WEAKNESS,
    Deck.Embedding.QUANTITY
])

# METRICHE 
metrics_to_evaluate = [
    # Similarity.COSENO,
    Similarity.JACCARD,
    Similarity.WEIGHTED_JACCARD
    # Similarity.OVERLAP,
    # Similarity.PEARSON
]

results = evaluate_clustering_with_embeddings(
    decks=decks,
    embeddings=embeddings_to_evaluate,
    metrics=metrics_to_evaluate,
    sample_size=1000
)
```

```python
df = results

# Colonne da usare
df["db_inverted"] = -df["davies_bouldin"]
cols = ["silhouette", "calinski_harabasz", "db_inverted"]

# Normalizza le metriche
scaler = MinMaxScaler()
df_norm = pd.DataFrame(scaler.fit_transform(df[cols]), columns=[c+"_norm" for c in cols])
df = pd.concat([df, df_norm], axis=1)

# Definisco i pesi delle metriche e lo score risultante
w1, w2, w3 = 1, 1, 1
df["score"] = (
    w1 * df["silhouette_norm"] +
    w2 * df["calinski_harabasz_norm"] +
    w3 * df["db_inverted_norm"]
)

# Ordina dal migliore al peggiore
df_sorted = df.sort_values("score", ascending=False)
df_sorted.to_csv("optimal_jaccard_clusters_resultss.csv", index=False)
```
